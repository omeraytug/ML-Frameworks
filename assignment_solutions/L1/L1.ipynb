```python
# Pytorch (eager)
import torch

def f(x):
    return x**3 + 2*x

x = torch.tensor(3.0)
y = f(x)
y.item()
```

```python
# Testa om maskinen har CUDA (NVIDIA) eller något annat

device = "cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu"
print(device)

# Att byta till cuda/mps (alltså grafikkortet), ger oss många fler beräkningsenheter -> mer effektiv beräkning
```

```python
# TensorFlow (graph)
import tensorflow as tf

@tf.function
def f_tf(x):
    return x**3 + 2*x

x_tf = tf.constant(3.0)
y_tf = f_tf(x_tf)

print("TensorFlow (graph) output:", y_tf.numpy())
```

```python
# JAX (graph-like)
import jax
import jax.numpy as jnp

def f_jax(x):
    return x**3 + 2*x

x_jax = jnp.array(3.0)
y_jax = f_jax(x_jax)

y_jax
```
