- TODO: Write a small function f(x) = x^3 + 2x
- TODO: Implement f(x) in ONE of:
    - PyTorch (eager)
    - TensorFlow with @tf.function (graph)
    - JAX with @jit (graph-like)
- TODO: Print the output and note how execution differs

```python
# Pytorch (eager)
import torch

def f_torch(x):
    return x**3 + 2*x

x_torch = torch.tensor(3.0)
y_torch = f_torch(x_torch)
y_torch.item()
```

```python
# TensorFlow (graph)
import tensorflow as tf

@tf.function
def f_tf(x):
    return x**3 + 2*x

x_tf = tf.constant(3.0)
y_tf = f_tf(x_tf)

print("TensorFlow (graph) output:", y_tf.numpy())
```

```python
# JAX (graph-like)
import jax
import jax.numpy as jnp

def f_jax(x):
    return x**3 + 2*x

x_jax = jnp.array(3.0)
y_jax = f_jax(x_jax)

y_jax
```
